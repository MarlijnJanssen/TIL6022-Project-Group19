## 1. Introduction

Car ownership is an important aspect of mobility and is closely related to car dependency. While cars provide travellers with accessibility and flexibility, high levels of car dependency are also associated with societal and environmental impacts. These include energy consumption, impacts on public health and the use of urban space (Sierra Muñoz et al., 2024). In the Netherlands, car ownership varies across regions. Private car ownership per resident has decreased in highly urbanised areas, while it has increased in more rural areas (Zijlstra et al., 2022). These differences are associated with household characteristics, such as income, urbanisation and the availability of alternative transport options (Zijlstra et al., 2022).

Public transport accessibility is one of the spatial factors that can be related to car ownership. Research in the Netherlands by Zijlstra et al. (2022) found that households living closer to railway stations generally own fewer cars, particularly near intercity stations. However, living close to public transport does not necessarily mean that households own fewer cars. Blomjous (2019) also found that households living further from railway stations generally owned more cars, even after accounting for socio-demographic and spatial characteristics. This association was relatively small compared to other factors influencing car ownership. The study also found differences between station types, with lower average car ownership in neighbourhoods near larger stations compared to smaller stations. Similarly, Chatman (2013) found lower car ownership and car use among households living near railway stations, but showed that factors such as parking availability, housing characteristics, density and bus services also played an important role. In the Netherlands, Debrezion et al. (2009) found that station choice depends not only on distance, but also on station characteristics such as service frequency and facilities. Railway stations also differ in their role within the railway network (Hansen, 2021). In this study, a distinction is made between major stations (intercity / junction hubs) and minor local stations. These findings suggest that the relationship between railway accessibility and car ownership may depend not only on station proximity, but also on the type of station and the service provided.

An important aspect of public transport service quality is reliability. Unreliable services can lead to longer or unpredictable waiting times and missed connections, which can make public transport less attractive to travellers. In the Netherlands, De Haas (2023) analysed changes in public transport use following the COVID-19 pandemic. The study found that some travellers identified unreliable train services as a reason for travelling less frequently by train. Reliability may therefore influence the attractiveness of rail transport as an alternative to car use. Railway disruptions, including delayed or cancelled train services, can provide an indication of railway reliability. Having a station nearby may therefore be less valuable when train services are frequently disrupted. However, while reliability has been studied in relation to public transport use, its role in the relationship between station proximity and car ownership is less clear.

This study examines the relationship between railway station proximity and car ownership across neighbourhoods in the Netherlands, while accounting for socio-economic and spatial differences between neighbourhoods. In addition, station type and railway disruptions are considered to examine whether this relationship differs depending on the type and reliability of the nearby station. The study aims to provide a better understanding of whether the association between station proximity and car ownership differs by station type and railway reliability. These insights can contribute to discussion on housing development near railway stations and the role of rail accessibility in reducing car dependency.

## 2. Research Objective

With this research, the relationship between railway station proximity and household car ownership is analysed in relation to certain socioeconomic and operational variables across Dutch neighbourhoods. One of the main objectives is to evaluate if living closer to a station decreases household car ownership, while controlling for income, urbanisation, household size, and population density. The research also determines whether this relationship differs between major intercity hubs and minor local stations, and whether it is weakened by frequent train disruptions. After analysing these spatial and statistical dependencies, the research concludes with an evaluation of the implications for station-oriented housing developments in the Netherlands.

Our main research question is: *Is there an association between distance to the nearest railway station and household car ownership in Dutch neighbourhoods (2023), and to what extent does this association vary by station type and by the share of disrupted trains at the nearest station (2023)?*

To get a comprehensive answer on the main research question, the following four sub-questions will be studied as well:

1. *How were average car ownership per household, distance to the nearest station and the share of disrupted trains at railway stations geographically distributed across the Netherlands in 2023?*
2. *What is the baseline association between station distance and household car ownership when accounting for income, urbanisation, household size and population density?*
3. *How does this association differ between major intercity hubs and minor local stations?*
4. *To what extent does frequent train disruption weaken the relationship between living near a station and lower household car ownership?*

## 3. Methodology

This research examines the relationship between railway proximity, train service reliability and household car ownership in the Netherlands. Datasets from Statistics Netherlands (CBS) and Rijden de Treinen were retrieved, processed and filtered. The analysis combines descriptive spatial exploration of neighbourhoods, spatial centroid matching of neighbourhoods to railway stations, aggregation of disruption features, and Ordinary Least Squares (OLS) multiple linear regression with interaction terms.

The analysis is done at neighbourhood (buurt) level, because this is the most detailed level at which open data on car ownership is available. The results therefore describe neighbourhoods, not individual households. OLS regression was chosen because the dependent variable, cars per household, is continuous. Also the coefficients can be interpreted directly as the change in cars per household per kilometre, while controlling for other neighbourhood characteristics. Income, urbanisation, household size and population density were included as control variables because the literature identifies them as main determinants of car ownership (Feyzollahi et al., 2024; Zijlstra et al., 2022). Without these control variables, the association with station distance would be overestimated. To answer SQ3 and SQ4, interaction terms were used instead of separate models per group, because an interaction term directly tests whether the distance association differs by station type or reliability.

### 3.2 Geographical and temporal scale

The study covers the whole of the Netherlands at neighbourhood (buurt) level, with 2023 as reference year.

- **Geographical scale:** all Dutch neighbourhoods (buurten) as defined by CBS in 2023, linked to all Dutch railway stations. Neighbourhoods without published car ownership figures are excluded.
- **Temporal scale:** a cross-sectional analysis for 2023. CBS neighbourhood figures, proximity statistics and the neighbourhood map all use the 2023 edition, so neighbourhood codes match. Disruption counts are averaged over 2021–2023 to smooth out one-off incidents; the share of disrupted trains is calculated for 2023.

2023 was chosen because it is the most recent year for which the CBS key figures, proximity statistics and neighbourhood map form a matching pair with identical spatial boundaries. Complete neighbourhood-level figures on car ownership and income are published by CBS with a lag of 12 to 18 months (CBS, 2024).

### 3.3 Known limitations of the data

To ensure data validity, several known dataset limitations were taken into account during data filtering:

- **Lease and rental cars.** CBS assigns cars by registration address. Cars registered at a lease or rental company inflate the figures for that neighbourhood (CBS, 2026). Such commercial, lease-heavy neighbourhoods were therefore excluded (see Phase 2).
- **Suppressed values.** Cars per household is only published for neighbourhoods with at least 50 households and at most 2.5 cars per household (CBS, 2024). Small rural neighbourhoods drop out.
- **Disruptions are assigned by line.** Affected stations are derived from the lines linked to a disruption, so stations elsewhere on those lines are also counted (Rijden de Treinen, 2026b). The disruption counts therefore represent general network disruption exposure rather than local, station-level disruptions.
- **Only NS-communicated disruptions.** A disruption is typically only communicated when multiple trains are affected; minor delays are not in the disruption data.
- **Two distance measures.** The nearest station is matched by straight-line distance from the neighbourhood centroid (Phase 3), while the distance variable in the models is the average road distance reported by CBS. These do not necessarily refer to the same station, so in some neighbourhoods the station type and share of disrupted trains may belong to a different station than the distance. The centroid is also not population-weighted, so it may not represent where most residents live.

## 4. Data pipeline

The data pipeline consists of seven phases, covering data acquisition, cleaning, integration and analysis. The datasets are combined into a single neighbourhood-level dataset, which is used for spatial visualisations and regression analyses.

### Phase 1: Data Acquisition and Initial Processing

The analysis uses data from Statistics Netherlands (CBS) and Rijden de Treinen. Two CBS datasets were obtained. The Kerncijfers wijken en buurten 2023 contains neighbourhood-level statistics on car ownership and socio-economic characteristics ([Kerncijfers wijken en buurten 2023](https://www.cbs.nl/nl-nl/cijfers/detail/85618NED); CBS, 2026), while Nabijheid voorzieningen 2023 provide information on the distance to facilities including railway stations ([Nabijheid voorzieningen; afstand locatie, wijk- en buurtcijfers 2023](https://www.cbs.nl/nl-nl/cijfers/detail/85830NED); CBS, 2025).

Railway station information, historical disruption records and train service data were obtained from Rijden de Treinen. The station dataset contains information on Dutch railway stations ([Open data about railway stations](https://www.rijdendetreinen.nl/en/open-data/stations); Rijden de Treinen, 2026a), while the disruption dataset provides historical records of railway disruptions ([Open data: train disruptions](https://www.rijdendetreinen.nl/en/open-data/disruptions); Rijden de Treinen, 2026b). The disruption records for 2021, 2022 and 2023 were combined into a single dataset containing 15,541 disruptions. In addition, the train archive for 2023 was obtained to calculate the percentage of delayed or cancelled trains at each station ([Open data about train services](https://www.rijdendetreinen.nl/en/open-data/train-archive); Rijden de Treinen, 2026c). The station code 'NA' (Nieuw Amsterdam) was restored because pandas initially interpreted it as a missing value.

The Wijk- en buurtkaart 2023 was loaded as a GeoDataFrame using the Dutch coordinate reference system (EPSG:28992) ([Wijk- en buurtkaart 2023, version 3](https://www.cbs.nl/nl-nl/dossier/nederland-regionaal/geografische-data/wijk-en-buurtkaart-2023); CBS & Kadaster, 2026). To prepare the CBS datasets for integration, trailing spaces were removed from the region codes. The region codes were then checked to ensure all region codes in the key figures dataset were also present in the proximity dataset. The original data files were stored separately to preserve the raw data and support reproducibility.

### Phase 2: Data Cleaning and Sample Selection

The relevant variables were selected from both CBS datasets and merged using the neighbourhood codes. Only neighbourhood-level records were retained, excluding observations at the national, municipal and district levels. Neighbourhoods with missing car ownership data were removed, as well as neighbourhoods with missing values for any of the variables used in the regression models: distance to the nearest railway station, income, urbanisation, household size and population density.

Additionally, neighbourhoods with more than one registered business per household were excluded. This threshold is approximately 3.5 times the national average of 0.29 businesses per household and was selected to identify commercial areas where company and lease cars may inflate the registered car ownership figures. The threshold was defined specifically for this study. After applying these filtering criteria, the dataset contained 11,221 neighbourhoods. The cleaned dataset was saved for use in the subsequent phases of the analysis.

### Phase 3: Spatial Station Matching

For each neighbourhood in the sample, the centroid was calculated using the CBS neighbourhood map. The railway stations were converted from latitude and longitude coordinates into geographic points and transformed to the same Dutch coordinate reference system as the neighbourhood map (EPSG:28992, in metres). Two event stations, Rotterdam Stadion and Heerenveen IJsstadion, were excluded because they do not provide regular daily train services. Each neighbourhood centroid was then linked to its nearest railway station based on straight-line distance. This added the station code, name and type to each neighbourhood.

Stations classified as megastations, intercity hubs or intercity stations were grouped as major stations, while all other stations were classified as minor (local) stations. Of the 11,221 neighbourhoods, 2,574 were linked to a major station and 8,647 to a minor station.

### Phase 4: Station Reliability

From the 15,541 recorded disruptions between 2021 and 2023, disruptions without station codes and planned engineering work were removed, leaving 14,928 disruptions. Since one disruption can affect multiple stations, each disruption was assigned to every affected station included in the Dutch station dataset.

For each station, the average annual number of disruptions and the average annual disruption duration in minutes were calculated by dividing the totals over the three year period by three. Stations without recorded disruptions were assigned a value of zero. The stations with the highest annual number of disruptions were Rotterdam Centraal (535), Schiphol Airport (433) and Utrecht Centraal (356).

In addition, the percentage of disrupted train stops was calculated using the 2023 train archive from Rijden de Treinen. The twelve monthly datasets were combined to determine the total number of train stops and disrupted stops at each station. A stop was classified as disrupted if the train was cancelled or arrived or departed at least five minutes late. The percentage of disrupted stops was calculated by dividing the number of disrupted stops by the total number of stops at each station and multiplying by 100. The average number of train stops per day was also calculated. The percentage of disrupted stops was used as the reliability measure in the regression analysis to account for differences in the number of trains serving each station.

### Phase 5: Master Dataset Integration

The neighbourhood dataset from Phase 3 was merged with the station reliability dataset from Phase 4 using the station code. This assigned the disruption figures of the nearest railway station to each neighbourhood. The resulting dataset contains 11,221 neighbourhoods with no missing values.

Next, neighbourhoods located more than 10 km from their nearest railway station were excluded. This filter was based on the road distance provided by CBS, rather than the straight-line distance used for station matching in Phase 3. The threshold was applied because station characteristics and reliability are less likely to be relevant to car ownership in neighbourhoods located further away. After applying the filter, the final dataset contained 9,039 neighbourhoods, representing 86.7% of the households in the dataset before the distance filter.

For each neighbourhood, the dataset includes car ownership per household (the dependent variable), the nearest station's code, name, type, size classification and the distance to the nearest railway station and nearest major transfer station. It also includes the average annual number and duration of disruptions over 2021-2023, the percentage of disrupted train stops in 2023 and the average number of train stops per day at the nearest station. The control variables are income, urbanisation, household size and population density. The final master dataset was saved for use in the subsequent spatial visualisations and regression analyses.

### Phase 6: Spatial and exploratory data analysis

The master dataset from Phase 5 was used to analyse the spatial distribution of household car ownership, station proximity and train disruptions across Dutch neighbourhoods in 2023. First, descriptive statistics were calculated for car ownership per household, distance to the nearest railway station and the percentage of disrupted trains.

The master dataset was then linked to the CBS neighbourhood map using neighbourhood codes. A national map was created showing car ownership per household, with major and minor railway stations displayed as separate markers. Neighbourhoods outside the selected sample were shown in light grey. The average car ownership of neighbourhoods linked to major and minor stations was also compared.

To analyse the relationship between station distance and car ownership, neighbourhoods were grouped into distance categories based on their distance to the nearest railway station. Categories of 1 km were used for distances up to 6 km, followed by two wider categories covering 6-8 km and 8-10 km. For each category, the average number of cars per household and the number of neighbourhoods were calculated. The average car ownership per distance category was visualised in a bar chart.

Finally, the percentage of disrupted trains in 2023 was linked to the railway station locations and displayed on a national map. The ten stations with the highest percentages of disrupted trains were also identified. These exploratory analyses were used to describe national patterns and answer subquestion 1 before conducting the regression analyses in Phase 7.

### Phase 7: Econometric and interaction modelling

The master dataset from Phase 5 was used to analyse the relationship between railway station proximity and household car ownership through Ordinary Least Squares (OLS) regression models. Before estimating the models, the explanatory variables were checked for multicollinearity using a correlation matrix and Variance Inflation Factors (VIF). Urbanisation was recoded so that higher values represented more urbanised neighbourhoods and population density was expressed in thousands of inhabitants per square kilometre to simplify the interpretation of the coefficients.

For subquestion 2, a multiple linear regression was estimated with car ownership per household as the dependent variable and distance to the nearest railway station as the main explanatory variable. Income, urbanisation, household size and population density were included as control variables. The explanatory variables were also standardised to compare their relative associations with car ownership. The standardised coefficients and their 95% confidence intervals were visualised in a coefficient plot.

For subquestion 3, the relationship between station distance and car ownership was compared between major and minor stations. First, separate scatter plots with linear trend lines were created for both station types without control variables. Subsequently, an interaction term between station distance and station type was added to the regression model, alongside the control variables, to test whether the association differed significantly between major and minor stations.

For subquestion 4, an interaction term between station distance and the percentage of disrupted trains was included in the regression model. Station type and the same socioeconomic and spatial control variables were also included. The association between station distance and car ownership was compared at low, median and high percentages of disrupted trains. A three dimensional surface plot was created to visualise predicted car ownership across different combinations of station distance and disruption percentage, while holding the control variables at their mean values.

Finally, two robustness checks were performed on the baseline model. The first added the distance to the nearest major transfer station as an additional explanatory variable. The second used standard errors clustered at the municipality level to account for similarities between neighbourhoods within the same municipality.

### 5.2 Baseline association between station distance and household car ownership (SQ2)

To analyse the association between station distance and household car ownership, an ordinary least squares (OLS) regression was estimated, controlling for income, urbanisation, household size and population density. A difference of 10 km therefore corresponds to approximately 0.05 additional cars per household. The model has an R² of 0.757, indicating that the included variables explain approximately 75.7% of the variation in household car ownership.

### 5.5 Robustness checks

Two robustness checks were performed to analyse whether the positive association between station distance and household car ownership found in sub-question 2 remained stable under different model specifications.

First, the distance to the nearest major transfer station was added to the baseline model. This was done to analyse whether the association with distance to the nearest station remained when also accounting for access to a major transfer station. The coefficient for distance to the nearest station remained positive and statistically significant (β = 0.0049, p < 0.001), compared with 0.0050 in the baseline model. Distance to the nearest major transfer station also showed a positive association with car ownership (β = 0.0025, p < 0.001). The R² increased slightly from 0.757 to 0.760. This indicates that the original association remained largely unchanged after including the additional distance variable.

Second, the baseline model was estimated using standard errors clustered at the municipality level. Neighbourhoods within the same municipality may share characteristics that make their regression errors related. Clustering the standard errors accounts for this possibility when assessing statistical significance. The coefficient for station distance remained 0.0050, while the p-value was 0.002. Therefore, the association remained statistically significant after accounting for possible similarities between neighbourhoods within municipalities. In addition, multicollinearity was analysed using variance inflation factors (VIF). This was done to check whether strong relationships between explanatory variables, such as urbanisation and population density, could affect the precision of the regression coefficients. The highest VIF was 2.72 for urbanisation, followed by 2.00 for population density. All VIF values were below the commonly used threshold of 5, indicating no problematic multicollinearity.

Overall, the robustness checks show that the positive association between station distance and household car ownership remained statistically significant. The estimated coefficient changed only slightly when an additional accessibility variable was included, and the association remained significant when standard errors were clustered by municipality. The VIF results also indicate that multicollinearity was not a major concern in the baseline model.

## 6. Conclusion

In this research, we examined whether distance to the nearest railway station is associated with household car ownership in Dutch neighbourhoods in 2023, and whether this differs by station type and rail reliability. Based on 9,039 neighbourhoods within 10 km of a railway station, the answer is that an association exists, but a small one. In the descriptive analysis, car ownership increased from 0.93 cars per household in neighbourhoods within 1 km of a station to 1.35 cars per household at 8 to 10 km. However, neighbourhoods close to stations are also more urban and more densely populated. After controlling for income, urbanisation, household size and population density, each additional kilometre to the nearest station was associated with only 0.005 more cars per household, about 0.05 over 10 km. This association stayed stable when distance to the nearest major transfer station was added and when standard errors were clustered per municipality. Station distance was a rather small predictor; household size, urbanisation and population density were much stronger predictors of car ownership.

The association did not differ between major stations and minor stations. Neighbourhoods near a major station owned slightly fewer cars overall (0.03 per household), but car ownership increased with distance at the same rate for both station types. Rail reliability also did not play a significant role. We measured this as the share of trains that were cancelled or delayed by at least five minutes in 2023. It did not weaken the association between station proximity and car ownership, and it was not associated with car ownership itself.

Overall, we conclude that living closer to a railway station is associated with lower household car ownership. However, this association is largely explained by the urban and dense character of the station areas rather than by the proximity itself. For housing development near railway stations, this suggests that station proximity alone is unlikely to substantially reduce car dependency of households. The density and urban character of the surrounding neighbourhood are more important.


## 7. Discussion

This study started from the idea that living close to a railway station could reduce the need for a car, and that this might depend on the type of station and on how reliable the train service is. In this section we explain what our results mean, how they relate to earlier research, how our choices in the data pipeline may have influenced, and what they imply for housing development near stations. 
The main lesson from our results is that the relationship between station proximity and car ownership looks much stronger at first sight than it actually is. Without control variables, car ownership clearly increased with distance to the station, but with control variables this effect became very small, and household size, urbanisation and population density were much stronger predictors. This is because stations are mostly located in towns and cities. Neighbourhoods close to a station are therefore also more urban, denser and have smaller households, and it is mainly these characteristics that go together with lower car ownership. Station proximity is thus part of a wider urban setting rather than a strong factor on its own.

The same explanation applies to station type. Without control variables, car ownership increased more steeply with distance near major stations, but with control variables the slopes were equal (p = 0.968). The steeper raw slope therefore comes from the large cities around major stations, not from the stations themselves. Only a small difference in level remained, with 0.031 fewer cars per household near major stations. This may be due to the better intercity services, but also to features of city centres that are not in our model, such as limited parking. Reliability did not change the picture either. A higher percentage of disrupted trains did not weaken the association with distance (p = 0.706). A likely reason is that owning a car is a long-term decision, and a few percentage points more delayed or cancelled trains is too small a difference to change it.

These findings are in line with earlier research. Blomjous (2019) also found that car ownership increases with distance to a station after controlling for other factors, but that this effect is small. Like Blomjous (2019) and Zijlstra et al. (2022), we found lower car ownership near larger stations, although in our case this was only a difference in level. Chatman (2013) found that lower car ownership near stations was explained more by parking, housing type and density than by rail access itself, which matches our results. Finally, De Haas (2023) found that unreliable trains make some people travel less by train. This does not contradict our results, as travelling less by train can happen quickly, while selling a car is a much bigger decision.

Several choices in our approach may have influenced these results. First, the distance comes from CBS and is measured by road, while the station type and reliability belong to the station closest in a straight line to the middle of the neighbourhood. In large neighbourhoods these can be two different stations. This does not affect the distance effect itself, but it adds noise to the station characteristics, which makes real differences between station types or reliability levels harder to detect. Second, we only used the nearest station, so the model does not show whether other stations are within a short distance. Neighbourhoods near two or more stations, for example in Utrecht or The Hague, have better rail access than our distance suggests, and travellers often choose a station further away when it has better services (Debrezion et al., 2009). When we added the distance to the nearest major transfer station, the distance effect hardly changed (0.0049 instead of 0.0050), so our main result does not seem to depend on this choice.

Third, in the disruption data a disruption is counted at every station on the affected line, so large stations on many lines score highest. This count was strongly related to the number of trains per day (correlation 0.83), so it mainly measured how busy a station was. We corrected for this by using the percentage of disrupted trains from the train archive, which divides the disrupted trains by all trains at a station and is hardly related to busyness (0.25). 
Fourth, we removed neighbourhoods with more than one business per household, because lease and company cars registered there inflate car ownership. This cut-off is about 3.5 times the national average, but it was our own choice and we did not test other values, so we cannot say how sensitive the results are to it.

Some important factors were also missing from our data, of which parking is probably the most important. Guo (2013) and Chatman (2013) both found that parking availability strongly affects car ownership. Neighbourhoods near stations, especially major stations, are often central areas with paid parking, permits and little street space. Part of what looks like a station effect may therefore be a parking effect. Finally, our data only show associations and not causal effects. Households that do not want a car may choose to live near a station on purpose (Cao et al., 2009).

Most of these limitations would make the true effect of the station smaller rather than larger. This supports our conclusion that building homes close to a station will not by itself lead to much lower car ownership. If the association were causal, 1,000 households next to a station would own only about 25 fewer cars than 1,000 similar households at 5 km. For station-area housing development, combining proximity with high density and a strict parking policy is therefore likely to be more effective than proximity alone. Future research could count all stations within a few kilometres of a neighbourhood, test other cut-offs for business areas, add data on parking and other public transport, and use data over several years, for example before and after a new station opens, to get closer to a causal effect.
