# Phase 5: Master dataset

In this notebook we combine the results of Phase 3 and Phase 4 into one table. Each row is one neighbourhood, with its car ownership, its nearest station, the station type, the number of disruptions at that station, and the control variables. This table is used for all maps and models.

```python
from pathlib import Path
import pandas as pd

PROCESSED = Path('data/cleaned_data')

# Load the results of Phase 3 and Phase 4
buurten_met_station = pd.read_pickle(PROCESSED / 'buurten_met_station.pkl')
storingen_per_station = pd.read_pickle(PROCESSED / 'storingen_per_station.pkl')

print(len(buurten_met_station))
print(len(storingen_per_station))
```

```python
# Add the disruption numbers to each neighbourhood, using the station code
master = buurten_met_station.merge(storingen_per_station, on='station_code', how='left')

# Check: still 11221 neighbourhoods, and no empty values
print(len(master))
print('Empty values:', master.isna().sum().sum())
```

```python
kolommen = [
    'buurtcode', 'buurtnaam', 'gemeente',          # which neighbourhood
    'autos_per_hh',                                 # what we want to explain
    'afstand_station', 'afstand_overstapstation',   # distance to the station
    'station_code', 'station_naam', 'station_type', 'station_groot',   # which station
    'storingen_per_jaar', 'minuten_per_jaar',       # how reliable the station is
    'inkomen', 'stedelijkheid', 'hh_grootte', 'dichtheid',             # control variables
]
master = master[kolommen]

master.head()
```
